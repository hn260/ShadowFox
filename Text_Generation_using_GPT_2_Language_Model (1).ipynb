# **Text Generation using GPT-2 Language Model**

Import Libraries



```python
from transformers import pipeline
import matplotlib.pyplot as plt
```

Load GPT-2 Model

```python
generator = pipeline("text-generation", model="gpt2")
```

Generate Text

```python
prompt = "Artificial Intelligence is"

result = generator(
    prompt,
    max_length=60,
    num_return_sequences=1
)

print(result[0]["generated_text"])
```

Test Multiple Prompts

```python
prompts = [
    "The future of education",
    "Machine learning is",
    "Python programming helps",
    "Climate change can"
]

for p in prompts:
    print("="*50)
    print("Prompt:", p)

    output = generator(
        p,
        max_length=50,
        num_return_sequences=1
    )

    print(output[0]["generated_text"])
```

Research Questions

```python
research_questions = [
    "Can GPT-2 complete sentences correctly?",
    "Can GPT-2 generate meaningful paragraphs?",
    "Does GPT-2 understand context?",
    "Is GPT-2 creative?"
]

for q in research_questions:
    print(q)
```

Simple Performance Analysis

```python
prompts = [
    "Artificial Intelligence",
    "Python",
    "Data Science",
    "Machine Learning"
]

lengths = []

for p in prompts:

    response = generator(
        p,
        max_length=50,
        num_return_sequences=1
    )

    text = response[0]["generated_text"]

    lengths.append(len(text.split()))

    print(text)
    print()
```

Visualization

```python
plt.bar(prompts, lengths)

plt.xlabel("Prompt")
plt.ylabel("Generated Words")
plt.title("GPT-2 Output Length")

plt.show()
```
